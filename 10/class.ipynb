## Tensor Basics

### Create Tensors
Tensors can be created by importing PyTorch and use `torch.tensor(data)`.

```python
import torch

t0 = torch.tensor(10) # 0D tensor
t1 = torch.tensor([1, 2, 3, 4]) # 1D tensor
t2 = torch.tensor([[1, 2, 3],
                   [4, 5, 6]]) # 2D tensor
print(t0)
print(t1)
print(t2)
print(t2.shape) # return the shape of the tensor
```

```python
# We can also use torch.zeros to create a tensor filled with zeros.
t = torch.zeros([3, 4])
print(t)
```

```python
# Ex: Create a 1D tensor filled with values from 1 to 10.

```

Another useful tool for representing multi-dimensional arrays is `numpy.array()`.

```python
import numpy as np
arr = np.array([[1, 2, 3],
                [4, 5, 6]])
arr
```

```python
t = torch.tensor(arr)
print(t)
```

Using `Tensor.numpy()` will convert a PyTorch tensor to a NumPy array.

```python
t.numpy()
```

### Reshaping Tensors
Sometimes we need to change the shape of the tensor.

```python
t5 = torch.tensor(range(12))
print(t5)
print(t5.shape)

t6 = t5.view(2, 6)
print(t6)
print(t6.shape)

# What if I want to create a tensor like this:
# [[0, 2, 4, 6, 8, 10],
#  [1, 3, 5, 7, 9, 11]]
print(torch.tensor(range(12)).view([6, 2]).T) # .T represents the transpose operation

t7 = t5.view(2, 2, 3)
# t7 = t5.view(3, 2, 2)
print(t7)

t8 = t5.view(1, 12)
print(t8)
print(t8.shape)
```

```python
# We can also use torch.reshape to rearrange the values.
t8 = t5.reshape(1, 12)
print(t8)
print(t8.shape)
```

```python
# We can use -1 as a placeholder
# t9 = t5.view(-1, 6)
t9 = t5.view(3, -1)
print(t9)

# Only one dimension can be represented with -1
# t = t5.view(2, -1, -1) # This will cause an error.
```

```python
# Add an additional dimension using `torch.unsqueeze()`
t = torch.unsqueeze(t5, 1)
print(t)
print(t.shape)
```

```python
t = torch.squeeze(t)
print(t)
```

### Indexing and Slicing Tensors
An individual value of a tensor is also a tensor:

```python
t5 = torch.tensor(range(12))
print(t5)
element = t5[5]
print(element)
```

If there is more than one dimension, then we need to provide an index for each dimension.

```python
t10 = torch.tensor(range(12)).view(2, 6)
print(t10)
print(t10[1, 5])
print(t10[0, 2:5]) # Extract multiple values
print(t10[0, :]) # : represents all feasible indices on this dimension
print(t10[0]) # this is a legal expression
```

### Arithmetic Operations on Tensors
Tensors support element-wise arithmetic operations.

```python
x = torch.tensor([1.2, 3.4, 5.6])
y = torch.tensor([9.7, 8.6, 7.5])
print(x + y)
print(y - x)
print(x * y)
print(x / y)
```

We can also perform arithmetic operations between a tensor and a number.

```python
z = torch.tensor(range(12)).reshape(3, 4)
print(z)

print(z + 5)

print(z / 10)
```

Tensors also support vector inner-products and matrix multiplications.

```python
u = torch.tensor([1, 2, 3])
v = torch.tensor([4, 5, 6])
u@v # or u.dot(v)
```

```python
A = torch.tensor([[1, 2, 3],
                  [4, 5, 6]])
B = torch.tensor([[11, 22],
                  [33, 44],
                  [55, 66]])
A@B # or torch.matmul(A, B)
```

### Moving Tensors to the GPU
By default, tensors are stored in memory associated with the **central processing unit (CPU)**. PyTorch tensors can also be stored on a **graphics processing unit (GPU)**. GPU is capable of performing massively parallel computations, which is often ideal for processing tensors.

<img src="https://www.researchgate.net/profile/Andrea-Manconi-2/publication/273383346/figure/fig7/AS:271520048087043@1441746832653/The-main-difference-between-CPUs-and-GPUs-is-related-to-the-number-of-available-cores-A.png" height="300">

```python
# Check if CUDA is available
torch.cuda.is_available()
```

```python
# Commonly-used method to represent the computaion device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print("Using device:", device)
```

```python
# Useful methods
print(torch.cuda.device_count())
print(torch.cuda.current_device())
print(torch.cuda.device(0))
print(torch.cuda.get_device_name(0))
print(torch.cuda.memory_allocated(0) / 1024 ** 3)
print(torch.cuda.memory_reserved(0) / 1024 ** 3)
```

```python
# Create a tensor on GPU
temp = torch.tensor([[4.0, 1.0],
                     [5.0, 3.0],
                     [2.0, 1.0]], device='cuda')
temp.device
```

```python
# Send an existing tensor to GPU
temp = torch.tensor([1, 2, 3, 4])
print(temp.device)

temp_gpu = temp.to('cuda') # or
temp_gpu = temp.to(device) # or
temp_gpu = temp.to(0)      # or
temp_gpu = temp.to('cuda:0')

temp_gpu.device
```

```python
# Tensors stored in the same device can be computed
temp2 = torch.tensor([5, 6, 7, 8])
print(temp2.device)

temp_gpu + temp2.to('cuda')
```

### Exercise: Perform Image Transformations with Tensors

```python
from urllib.request import urlretrieve
from PIL import Image

url = "https://www.si.edu/sites/default/files/blog/scta-copy1.jpg" # scarlet tanager
urlretrieve(url, 'bird.jpg')
img = Image.open("bird.jpg")
img
```

```python
from torchvision import transforms
preprocess = transforms.Compose([
        transforms.Resize(256),
        transforms.ToTensor(),
        ])
```

```python
# The original img object is not a tensor
img.shape
```

```python
img_processed = preprocess(img)
print(img_processed.shape)
to_image = transforms.ToPILImage()
to_image(img_processed)
```

```python
# Visualize three color channels: 0 - Red, 1 - Green, 2 - Blue
img_copy = img_processed.clone()
img_copy[0, :, :] = 0
img_copy[2, :, :] = 0
to_image(img_copy)
```

```python
# Horizontal Flip
flip = transforms.RandomHorizontalFlip(p=1.0)
to_image(flip(img_processed))
```

```python
# Ex: Perform horizontal flip explicitly.

```

```python
# Perform center crop: Keep the center 224 * 224 pixels
crop = transforms.Compose([transforms.CenterCrop(224)])
to_image(crop(img_processed))
```

```python
# Ex: Perform the center crop explicitly.

```

```python
# Padding
pad = transforms.Compose([transforms.Pad(20)])
result = pad(img_processed)
print(result.shape)
# print("Upper-left block:")
# print(result[:5, :5])
to_image(pad(img_processed))
```

```python
# Ex: Perform padding explicitly.

```

```python
# Greyscale
grey = transforms.Compose([transforms.Grayscale(num_output_channels=1)])
to_image(grey(img_processed))
```

```python
# Ex: Perform greyscaling using the formula:
# 0.299 ∙ Red + 0.587 ∙ Green + 0.114 ∙ Blue

```
